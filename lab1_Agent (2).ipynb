```python
from langchain_groq import ChatGroq
from langchain_core.tools import tool
```

```python
import os
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="llama-3.1-8b-instant",
    api_key=os.environ["GROQ_API_KEY"],
    temperature=0,
)

print("Groq Connected Successfully!")
```

```python
import wikipedia

@tool
def wikipedia_search(query: str) -> str:
    """Search Wikipedia for factual information."""
    try:
        return wikipedia.summary(query, sentences=3)
    except Exception as e:
        return str(e)
```

```python
@tool
def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""

    allowed = set("0123456789+-*/(). ")

    if not set(expression).issubset(allowed):
        return "Invalid expression"

    try:
        return str(eval(expression))
    except Exception as e:
        return str(e)
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[wikipedia_search, calculator],
    system_prompt="""
You are a helpful AI assistant.

Use the Wikipedia tool whenever factual information is needed.

Use the Calculator tool whenever mathematics is required.

Always use tools before answering.
"""
)
```

```python
response = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content":
                "Who developed the theory of relativity and what is his birth year multiplied by 2?"
            }
        ]
    }
)

print(response["messages"][-1].content)
```
