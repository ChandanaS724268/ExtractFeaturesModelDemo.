```python
import os
from google.colab import userdata

os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0.3,
    api_key=os.environ["GROQ_API_KEY"]
)

print("Groq model connected successfully!")
```

```python
from langchain_community.tools import DuckDuckGoSearchResults

search = DuckDuckGoSearchResults(output_format="string")

def search_agent(topic):
    return search.run(f"latest news about {topic}")
```

```python
def summariser_agent(raw_results):

    prompt = f"""
You are a news summariser.

Summarise the following search results into exactly 3 concise bullet points.

Search Results:

{raw_results}
"""

    response = llm.invoke(prompt)

    return response.content
```

```python
import os
from google.colab import userdata

os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

print("API Key Loaded:", "GROQ_API_KEY" in os.environ)
```

```python
from langchain_groq import ChatGroq
import os

llm = ChatGroq(
    model="llama-3.1-8b-instant",
    api_key=os.environ["GROQ_API_KEY"],
    temperature=0.3
)

print("LLM Connected Successfully!")
```

```python
response = llm.invoke("Say hello.")

print(response.content)
```

```python
summary = summariser_agent(search_agent("Artificial Intelligence"))
print(summary)
```

```python
def report_agent(topic, summary):

    prompt = f"""
Create a Markdown report.

Title:
# News: {topic}

Include:

- Heading
- 3 bullet points
- One concluding sentence
"""

    prompt += "\n\nSummary:\n" + summary

    response = llm.invoke(prompt)

    return response.content
```

```python
topic = "Artificial Intelligence"

print("Searching news...")
raw = search_agent(topic)

print("Summarising...")
summary = summariser_agent(raw)

print("Generating report...")
report = report_agent(topic, summary)

print(report)
```

```python
with open("news_report.md", "w", encoding="utf-8") as f:
    f.write(report)

print("Report saved successfully!")
```
